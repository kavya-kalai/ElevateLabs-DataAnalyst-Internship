```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cleaned_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
```

```python
store_sales = df.groupby('Store')['Weekly_Sales'].sum().sort_values(ascending=False).head(10)
store_sales.plot(kind='bar', color='skyblue', figsize=(10,6))
plt.title('Top 10 Stores by Total Sales')
plt.ylabel('Total Sales')
plt.show()
```

Insight 1: "The line chart shows a massive spike in sales during late December, likely due to holiday shopping."

```python
plt.hist(df['Temperature'], bins=20, color='orange', edgecolor='black')
plt.title('Distribution of Temperature')
plt.xlabel('Temperature')
plt.show()
```

Insight 2: "The bar chart identifies Store X as the top performer, contributing significantly more revenue than the bottom 5 stores combined."

```python
plt.scatter(df['Fuel_Price'], df['Weekly_Sales'], alpha=0.5)
plt.title('Fuel Price vs. Weekly Sales')
plt.xlabel('Fuel Price')
plt.ylabel('Sales')
plt.show()
```

Insight 3: "The scatter plot shows no strong correlation between fuel prices and sales, suggesting that fuel costs do not immediately deter shoppers."
